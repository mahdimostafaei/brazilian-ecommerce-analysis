```python
# distribution of rows per prefix
geo_loc["geolocation_zip_code_prefix"].value_counts().head(10)
```

```python
# coordinate range sanity - lat between -34, 6
Brazil_lat = geo_loc["geolocation_lat"].min(), geo_loc["geolocation_lat"].max()
display(Brazil_lat)
```

```python
# coordinate range sanity - lng between -75, -30
Brazil_lng = geo_loc["geolocation_lng"].min(), geo_loc["geolocation_lng"].max()
display(Brazil_lng)
```

```python
#the range of lat & lng is outside Brazil. First we check if they are just outliers or not, if percentage is low we ignore.

# 1. Define the mask for outliers
outlier_mask = (
    (geo_loc["geolocation_lat"] < -37) | (geo_loc["geolocation_lat"] > 5) |
    (geo_loc["geolocation_lng"] < -75) | (geo_loc["geolocation_lng"] > -30)
)

# 2. Count them
outlier_count = outlier_mask.sum()

# 3. Calculate percentage
outlier_pct = (outlier_count / len(geo_loc)) * 100

print(f"Total outlier rows: {outlier_count}")
print(f"Percentage of total data: {outlier_pct:.4f}%")
```
